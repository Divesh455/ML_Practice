```python
total = df['play'].value_counts().sum()
Y = df['play'].value_counts().iloc[0]
N = df['play'].value_counts().iloc[1]
```

```python
PY = Y / total
PN = N / total
```

```python
# Outlook

Total_NO = pd.crosstab(df['outlook'], df['play']).sum()['No']
Total_Yes = pd.crosstab(df['outlook'], df['play']).sum()['Yes']

Overcast_No = pd.crosstab(df['outlook'], df['play']).loc['Overcast', 'No']

Rain_No = pd.crosstab(df['outlook'], df['play']).loc['Rain', 'No']

Sunny_No = pd.crosstab(df['outlook'], df['play']).loc['Sunny', 'No']

Overcast_Yes = pd.crosstab(df['outlook'], df['play']).loc['Overcast', 'Yes']

Rain_Yes = pd.crosstab(df['outlook'], df['play']).loc['Rain', 'Yes']

Sunny_Yes = pd.crosstab(df['outlook'], df['play']).loc['Sunny', 'Yes']
```

```python
PON = Overcast_No / Total_NO
PRN = Rain_No / Total_NO
PSN = Sunny_No / Total_NO

POY = Overcast_Yes / Total_Yes
PRY = Rain_Yes / Total_Yes
PSY = Sunny_Yes / Total_Yes
```

```python
# TEMP

Total_T_NO = pd.crosstab(df['temp'], df['play']).sum()['No']
Total_T_Yes = pd.crosstab(df['temp'], df['play']).sum()['Yes']

Hot_No = pd.crosstab(df['temp'], df['play']).loc['Hot', 'No']

Mild_No = pd.crosstab(df['temp'], df['play']).loc['Mild', 'No']

Cool_No = pd.crosstab(df['temp'], df['play']).loc['Cool', 'No']

Hot_Yes = pd.crosstab(df['temp'], df['play']).loc['Hot', 'Yes']

Mild_Yes = pd.crosstab(df['temp'], df['play']).loc['Mild', 'Yes']

Cool_Yes = pd.crosstab(df['temp'], df['play']).loc['Cool', 'Yes']
```

```python
PHN = Hot_No / Total_T_NO
PMN = Mild_No / Total_T_NO
PCN = Cool_No / Total_T_NO

PHY = Hot_Yes / Total_T_Yes
PMY = Mild_Yes / Total_T_Yes
PCY = Cool_Yes / Total_T_Yes
```

```python
# Humidity


Total_H_NO = pd.crosstab(df['humidity'], df['play']).sum()['No']
Total_H_Yes = pd.crosstab(df['humidity'], df['play']).sum()['Yes']

High_No = pd.crosstab(df['humidity'], df['play']).loc['High', 'No']

Normal_No = pd.crosstab(df['humidity'], df['play']).loc['Normal', 'No']



High_Yes = pd.crosstab(df['humidity'], df['play']).loc['High', 'Yes']

Normal_Yes = pd.crosstab(df['humidity'], df['play']).loc['Normal', 'Yes']

```

```python
PHHN = High_No / Total_H_NO
PNN = Normal_No / Total_H_NO


PHHY = High_Yes / Total_H_Yes
PNY = Normal_Yes / Total_H_Yes
```

```python
# wind

Total_W_NO = pd.crosstab(df['wind'], df['play']).sum()['No']
Total_W_Yes = pd.crosstab(df['wind'], df['play']).sum()['Yes']

Weak_No = pd.crosstab(df['wind'], df['play']).loc['Weak', 'No']

Strong_No = pd.crosstab(df['wind'], df['play']).loc['Strong', 'No']



Weak_Yes = pd.crosstab(df['wind'], df['play']).loc['Weak', 'Yes']

Strong_Yes = pd.crosstab(df['wind'], df['play']).loc['Strong', 'Yes']
```

```python
PWN = Weak_No / Total_W_NO
PSN = Strong_No / Total_W_NO

PWY = Weak_Yes / Total_W_Yes
PSY = Strong_Yes / Total_W_Yes
```

```python
# For OutLook = Sunny, Temp=HOt, Humadity = High , Wind = Weak

PYes = PY * PSY * PHY * PHHY * PWY
PNo = PN * PSN * PHN * PHHN * PWN
```
